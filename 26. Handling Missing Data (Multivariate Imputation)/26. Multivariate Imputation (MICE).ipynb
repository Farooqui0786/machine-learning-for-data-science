```python
# import pandas library for data manipulation
import pandas as pd

# import numpy library for numerical operations
import numpy as np

# import linear regression model for predicting continuous values
from sklearn.linear_model import LinearRegression
```

```python
# read dataset, select required columns, scale values, and round them
df = np.round(pd.read_csv('50_Startups.csv')[['R&D Spend','Administration','Marketing Spend','Profit']]/10000)

# set random seed for reproducible sampling
np.random.seed(9)

# randomly select 5 rows from the dataset
df = df.sample(5)

# display the sampled dataframe
df
```

```python
# remove the target column and keep only input feature columns
df = df.iloc[:,0:-1]

# display the updated dataframe
df
```

```python
# insert a missing value in the first feature of the second row
df.iloc[1,0] = np.nan

# insert a missing value in the second feature of the fourth row
df.iloc[3,1] = np.nan

# insert a missing value in the last feature of the last row
df.iloc[-1,-1] = np.nan
```

```python
# display first five rows of the dataframe
df.head()
```

```python
# create a new dataframe to store imputed values
df0 = pd.DataFrame()

# fill missing values in R&D Spend with its column mean
df0['R&D Spend'] = df['R&D Spend'].fillna(df['R&D Spend'].mean())

# fill missing values in Administration with its column mean
df0['Administration'] = df['Administration'].fillna(df['Administration'].mean())

# fill missing values in Marketing Spend with its column mean
df0['Marketing Spend'] = df['Marketing Spend'].fillna(df['Marketing Spend'].mean())
```

```python
# display dataframe after completing the first imputation iteration
df0
```

```python
# create a copy of the imputed dataframe for the next iteration
df1 = df0.copy()

# restore the original missing value in the first column of the second row
df1.iloc[1,0] = np.nan

# display the updated dataframe
df1
```

```python
# select rows without missing target value and use remaining columns as features
X = df1.iloc[[0,2,3,4],1:3]

# display the feature dataset
X
```

```python
# select corresponding values of the first column as target variable
y = df1.iloc[[0,2,3,4],0]

# display the target variable
y
```

```python
# create linear regression model object
lr = LinearRegression()

# train the model using selected features and target values
lr.fit(X,y)

# predict the missing value using the trained model
lr.predict(df1.iloc[1,1:].values.reshape(1,2))
```

```python
# replace the missing value with the predicted value
df1.iloc[1,0] = 23.14
```
