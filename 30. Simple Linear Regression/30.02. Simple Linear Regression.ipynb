```python
class MeraLR:

    # initialize the model parameters
    def __init__(self):
        self.m = None
        self.b = None

    # train the model and calculate slope and intercept
    def fit(self, X_train, y_train):

        # initialize numerator and denominator for calculating slope
        num = 0
        den = 0

        # calculate the numerator and denominator for the slope
        for i in range(X_train.shape[0]):

            # calculate the covariance between X and y
            num = num + ((X_train[i] - X_train.mean()) * (y_train[i] - y_train.mean()))

            # calculate the variance of X
            den = den + ((X_train[i] - X_train.mean()) * (X_train[i] - X_train.mean()))

        # calculate the slope using the formula m = covariance / variance
        self.m = num / den

        # calculate the intercept using the formula b = mean(y) - m * mean(x)
        self.b = y_train.mean() - (self.m * X_train.mean())

        # display the calculated slope
        print(self.m)

        # display the calculated intercept
        print(self.b)

    # predict the target values for new input data
    def predict(self, X_test):

        # display the input test values
        print(X_test)

        # calculate predictions using the equation y = mx + b
        return self.m * X_test + self.b
```
