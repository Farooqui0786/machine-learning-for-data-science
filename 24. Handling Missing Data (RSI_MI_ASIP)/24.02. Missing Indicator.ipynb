```python
# import NumPy for numerical operations
import numpy as np

# import Pandas for data manipulation
import pandas as pd

# import train_test_split to split the dataset
from sklearn.model_selection import train_test_split

# import MissingIndicator to create missing value indicator columns
# import SimpleImputer to fill missing values
from sklearn.impute import MissingIndicator, SimpleImputer
```

```python
# load only the required columns from the CSV file
df = pd.read_csv(
    'train.csv',
    usecols=['Age', 'Fare', 'Survived']
)
```

```python
# show the first 5 rows of the dataset
df.head()
```

```python
# create the input features by removing the target column
X = df.drop(columns=['Survived'])

# store the target column separately
y = df['Survived']
```

```python
# split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,      # use 20% of the data for testing
    random_state=2      # ensure the same split every time
)
```

```python
# show the first 5 rows of the training dataset
X_train.head()
```

```python
# create a SimpleImputer object with the default strategy
si = SimpleImputer()

# learn the imputation values from the training data and transform it
X_train_trf = si.fit_transform(X_train)

# apply the same imputation to the testing data
X_test_trf = si.transform(X_test)
```

```python
# display the transformed training data
X_train_trf
```

```python
# import the Logistic Regression model
from sklearn.linear_model import LogisticRegression

# create a Logistic Regression model
clf = LogisticRegression()

# train the model using the transformed training data
clf.fit(X_train_trf, y_train)

# predict the target values for the testing data
y_pred = clf.predict(X_test_trf)

# import the accuracy evaluation metric
from sklearn.metrics import accuracy_score

# calculate the model accuracy
accuracy_score(y_test, y_pred)
```

## **Using Missing Imputer**

```python
# create a MissingIndicator object
mi = MissingIndicator()

# learn which columns contain missing values in the training data
mi.fit(X_train)
```

```python
# display the indices of the columns that contain missing values
mi.features_
```

```python
# create missing value indicator columns for the training data
X_train_missing = mi.transform(X_train)
```

```python
# display the missing value indicator array
X_train_missing
```

```python
# create missing value indicator columns for the testing data
X_test_missing = mi.transform(X_test)
```

```python
# display the missing value indicator array for the testing data
X_test_missing
```

```python
# add the missing value indicator as a new column
X_train['Age_NA'] = X_train_missing
```

```python
# display the training dataset
X_train
```

```python
# add the missing value indicator as a new column in the testing data
X_test['Age_NA'] = X_test_missing
```

```python
# display the test dataset
X_test
```

```python
# create simple imputer object with default strategy to replace missing values with mean
si = SimpleImputer()

# fit imputer on training data and transform training data by filling missing values
X_train_trf2 = si.fit_transform(X_train)

# transform testing data using values learned from training data
X_test_trf2 = si.transform(X_test)
```

```python
# import logistic regression algorithm for classification
from sklearn.linear_model import LogisticRegression

# create logistic regression model object
clf = LogisticRegression()

# train the model using transformed training data
clf.fit(X_train_trf2,y_train)

# predict target values for transformed test data
y_pred = clf.predict(X_test_trf2)

# import accuracy score metric to evaluate model performance
from sklearn.metrics import accuracy_score

# calculate accuracy of the model predictions
accuracy_score(y_test,y_pred)
```
