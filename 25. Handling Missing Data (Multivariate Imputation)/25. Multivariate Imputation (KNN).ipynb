```python
# import numpy library for numerical operations
import numpy as np

# import pandas library for data manipulation
import pandas as pd

# import train test split function to divide dataset into training and testing sets
from sklearn.model_selection import train_test_split

# import knn imputer and simple imputer to handle missing values
from sklearn.impute import KNNImputer, SimpleImputer

# import logistic regression algorithm for classification
from sklearn.linear_model import LogisticRegression

# import accuracy score metric to evaluate model performance
from sklearn.metrics import accuracy_score
```

```python
# read csv file and select required columns from the dataset
df = pd.read_csv('train.csv')[['Age','Pclass','Fare','Survived']]
```

```python
# display first five rows of the dataset
df.head()
```

```python
# calculate percentage of missing values in each column
df.isnull().mean() * 100
```

```python
# separate input features by removing target column from dataset
X = df.drop(columns=['Survived'])

# store target variable for prediction
y = df['Survived']
```

```python
# split dataset into training and testing sets for model training and evaluation
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
# display first five rows of the training feature dataset
X_train.head()
```

```python
# create knn imputer object with 3 nearest neighbors and distance based weighting
knn = KNNImputer(n_neighbors=3,weights='distance')

# fit knn imputer on training data and fill missing values
X_train_trf = knn.fit_transform(X_train)

# fill missing values in test data using learned neighbors from training data
X_test_trf = knn.transform(X_test)
```

```python
# create logistic regression model object
lr = LogisticRegression()

# train the model using transformed training data
lr.fit(X_train_trf,y_train)

# predict target values for transformed test data
y_pred = lr.predict(X_test_trf)

# calculate accuracy of the model predictions
accuracy_score(y_test,y_pred)
```

### **Comparision with Simple Imputer**

```python
# create simple imputer object with default mean strategy for comparison
si = SimpleImputer()

# fit simple imputer on training data and fill missing values
X_train_trf2 = si.fit_transform(X_train)

# fill missing values in test data using values learned from training data
X_test_trf2 = si.transform(X_test)
```

```python
# create logistic regression model object
lr = LogisticRegression()

# train the model using data processed with simple imputer
lr.fit(X_train_trf2,y_train)

# predict target values for transformed test data
y_pred2 = lr.predict(X_test_trf2)

# calculate accuracy of the model predictions
accuracy_score(y_test,y_pred2)
```

Comparing with SimpleImputer, we see that th e KNNImputer gives better results
