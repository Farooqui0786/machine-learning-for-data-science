```python
# import a function to generate a synthetic regression dataset
from sklearn.datasets import make_regression

# import libraries for data manipulation and numerical operations
import pandas as pd
import numpy as np

# import Plotly libraries for interactive visualizations
import plotly.express as px
import plotly.graph_objects as go

# import regression evaluation metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# generate a synthetic regression dataset
X, y = make_regression(
    n_samples=100,      # create 100 data points
    n_features=2,       # create 2 input features
    n_informative=2,    # both features contribute to the target
    n_targets=1,        # create one target variable
    noise=50            # add random noise to the target values
)
```

```python
# create a DataFrame using the two features and the target variable
df = pd.DataFrame({
    'feature1': X[:, 0],
    'feature2': X[:, 1],
    'target': y
})
```

```python
# display the number of rows and columns in the DataFrame
df.shape
```

```python
# display the first five rows of the DataFrame
df.head()
```

```python
# create a 3D scatter plot using the two features and target
fig = px.scatter_3d(
    df,
    x='feature1',
    y='feature2',
    z='target'
)

# display the interactive 3D plot
fig.show()
```

```python
# import the function to split the dataset into training and testing sets
from sklearn.model_selection import train_test_split

# split the data into 80% training and 20% testing sets
# random_state=3 ensures the same split every time
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=3
)
```

```python
# import the linear regression model
from sklearn.linear_model import LinearRegression

# create an object of the linear regression model
lr = LinearRegression()
```

```python
# train the linear regression model using the training data
lr.fit(X_train, y_train)
```

```python
# make predictions on the test data using the trained linear regression model
y_pred = lr.predict(X_test)
```

```python
# calculate the mean absolute error
print("MAE:", mean_absolute_error(y_test, y_pred))

# calculate the mean squared error
print("MSE:", mean_squared_error(y_test, y_pred))

# calculate the R-squared score
print("R2 score:", r2_score(y_test, y_pred))
```

```python
# create 10 evenly spaced values between -5 and 5 for the x-axis
x = np.linspace(-5, 5, 10)

# create 10 evenly spaced values between -5 and 5 for the y-axis
y = np.linspace(-5, 5, 10)

# create a 2D grid from the x and y values
xGrid, yGrid = np.meshgrid(y, x)

# combine the grid coordinates into a 2D array of feature values
final = np.vstack((
    xGrid.ravel().reshape(1, 100),
    yGrid.ravel().reshape(1, 100)
)).T

# predict the target value for every point in the grid
# reshape the predictions into a 10x10 matrix for surface plotting
z_final = lr.predict(final).reshape(10, 10)

# store the predicted values in z
z = z_final
```

```python
# create a 3D scatter plot of the actual data points
fig = px.scatter_3d(
    df,
    x='feature1',
    y='feature2',
    z='target'
)

# add the linear regression plane to the 3D plot
fig.add_trace(
    go.Surface(
        x=x,
        y=y,
        z=z
    )
)

# display the interactive 3D plot
fig.show()
```
